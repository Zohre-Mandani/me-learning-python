hospitalized patients , blood sugar and blood pressure

```python
names = []
blood_pressures = []
blood_sugars = []
statuses = []


def menu_help():
    print("options")
    print("1 or add: add a new patient")
    print("2 or display: show all admitted patients")
    print("3 or edite: change patient info")
    print("4 or search: search patient by name")
    print("5 or exit:Exite game")


def existing_user():
    print("user already exist")


def adding_data(name, blood_pressure, blood_sugar, status):
    names.append(name)
    blood_pressures.append(blood_pressure)
    blood_sugars.append(blood_sugar)
    statuses.append(status)


def status_bp_bs(blood_pressure, blood_sugar):
    if blood_pressure > 140 and blood_sugar > 180:
        status = "high BP / high BS"
        return status
    elif blood_pressure > 140 and blood_sugar <= 180:
        status = "high BP"
        return status
    elif blood_sugar > 180 and blood_pressure <= 120:
        status = "high BS"
        return status
    print("No hospitalisation needed")
    return -1


def display(names):
    for a in range(len(names)):
        print(
            f"patient{a+1} / name: {names[a]} / BP: {blood_pressures[a]} / BS: {blood_sugars[a]} / status: {statuses[a]}")
    if len(names) == 0:
        print("No hospitalised patients!")


def no_patient_x(name):
    print(f"patient ({name}) not found!")


def edit_patient_data(name, blood_pressure, blood_sugar):
    i = names.index(name)
    blood_pressures[i] = blood_pressure
    blood_sugars[i] = blood_sugar
    statuses[i] = status_bp_bs(blood_pressure, blood_sugar)


def search_info(name):
    i = names.index(name)
    task = input(f"what do you need about patient{name}?blood pressure or blood sugar or status?").lower().strip()
    if task == "blood pressure":
        print(f"BP={blood_sugars[i]}")
    elif task == "blood sugar":
        print(f"BS={blood_sugars[i]}")
    elif task == "status":
        print(f"status:{statuses[i]}")
    else:
        print(f"There is no data called {task} for patient")


while True:
    choice = input(
        "choose 1-5, or print 'help': ").strip().lower()

    if choice == "help":
        menu_help()
    elif choice == "1" or choice == "add":
        name = input("Enter new patient's name").lower().strip()
        if name in names:
            existing_user()
        else:
            blood_pressure = int(input("blood pressure=?"))
            blood_sugar = int(input("blood sugar=?"))
            status = status_bp_bs(blood_pressure, blood_sugar)
            if status != -1:
                adding_data(name, blood_pressure, blood_sugar, status)
    elif choice == "2" or choice == "display":
        display(names)
    elif choice == "3" or choice == "edit":
        name = input("who's info do you want to edite?").lower().strip()
        if name in names:
            u = names.index(name)
            blood_pressure = int(input("enter new blood pressure"))
            blood_sugar = int(input("enter new blood sugar"))
            edit_patient_data(name, blood_pressure, blood_sugar)
            if statuses[u] == -1:
                names.pop(u)
                blood_pressures.pop(u)
                blood_sugars.pop(u)
                statuses.pop(u)
        else:
            no_patient_x(name)

    elif choice == "4" or choice == "search":
        name = input("wich user are you looking for?").lower().strip()
        if name in names:
            search_info(name)
            
        else:
            no_patient_x(name)

    elif choice == "5" or choice == "exit":
        break
    else:
        print("The command cannot be executed")
        print("Please try another command")
        
```
