Directory organizer

```python
# 1
import os
import shutil
# enter path
main_path = input("enter the main path")


# find files

def finding_files(path):
    file_pn = []
    for R, P, F in os.walk(path):
        for f in F:
            f_path = os.path.join(R, f)
            file_pn.append(f_path)
    return file_pn


def find_files_ext(files):
    exts = set()
    for f in files:
        ext = f.split(".")
        exts.add(ext[-1])
    return exts


def create_orgnize(m_path, files, exts):
    for f in files:
        file_name = os.path.basename(f)
        old_path = f
        ext = file_name.split(".")[-1]
        for e in exts:
            if ext == e:
                folder_name = "folder"+"_"+e
                new_path = os.path.join(m_path, folder_name)
                if not os.path.exists(new_path):
                    os.makedirs(new_path)
                    print(f"{folder_name} created in {new_path}")
                new_path = os.path.join(file_name, new_path)

                # print(file_name)
                try:
                    shutil.move(old_path, new_path)
                except Exception as e:
                    print(f"{file_name}--/--> {new_path}", e)


all_files = finding_files(main_path)
exts_set = find_files_ext(all_files)
create_orgnize(main_path, all_files, exts_set)
```
